This notebook inspects the real-world UCI Statlog German Credit dataset and the controlled synthetic benchmark.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load_german_credit, generate_controlled_synthetic

# 1. Load German Credit Data
df_german, report_german = load_german_credit()
print("German Credit Shape:", df_german.shape)
display(df_german.head())
```

```python
# 2. Target Class Distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df_german, x='target', palette=['#E53935', '#43A047'])
plt.title('Target Outcome Distribution: Good (1) vs Bad (0) Credit Risk')
plt.xlabel('Credit Decision')
plt.ylabel('Applicant Count')
plt.xticks([0, 1], ['Rejected (0)', 'Approved (1)'])
plt.show()
```
